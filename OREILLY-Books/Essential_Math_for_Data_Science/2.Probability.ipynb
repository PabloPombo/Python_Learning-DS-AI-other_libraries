### Binomial Distribution 

```python
#How likely k success can happen out of n trials given p probability 
```

```python
from scipy.stats import binom
```

```python
n = 10
p = 0.9

for k in range(n + 1):
    probability = binom.pmf(k, n, p)
    print("{0} - {1}".format(k,probability))
```

### Beta Distribution

```python
##Probability of probabilities ------- Meta-probability
```

```python
from scipy.stats import beta
```

```python
a = 8
b = 2

p = beta.cdf(0.90, a, b)
print(p)

#There is 77.48% chance to underlying probability of success is 90% or less
```

```python
a = 8
b = 2

p = 1.0 - beta.cdf(0.90, a, b)
print(p)

#There is only 22.5% chance the underlying success rate is 90% greater.
```

### Binomial distribution from Scratch 

```python
#Factorial multiply consecutive decending integers down to 1 
def factorial(n: int):
    f = 1
    for i in range(n):
        f *= (i + 1)
    return f

#Generates the coefficient needed for the binomial distribution 
def binomial_coefficient(n: int, k: int):
    return factorial(n) / (factorial(k) * factorial(n - k))

#Binomial distribution calculates the probability ofk events out of n trials 
#give the p probability of k ocurring
def binomial_distribution(k: int, n:int, p:float):
    return binomial_coefficient(n, k) * (p ** k) * (1.0 - p) ** (n - k)

#10 trials where each has 90% success probability
n=10
p=0.9

for k in range (n + 1):
    probability = binomial_distribution(k,n,p)
    print("{0} - {1}".format(k, probability))
```

### Beta distribution from Scratch 

```python
#Factorial multiply consecutive decending integers down to 1 
def factorial(n: int):
    f = 1
    for i in range(n):
        f *= (i + 1)
    return f

def approximate_integral(a, b, n, f):
    delta_x = (b - a) / n
    total_sum = 0
    
    for i in range(1, n + 1):
        midpoint = 0.5 * (2 * a + delta_x * (2 * i - 1))
        total_sum += f(midpoint)
        
    return total_sum * delta_x

def beta_distribution(x: float, alpha: float, beta: float) -> float:
    if x < 0.0 or x > 1.0:
        raise ValueError("x must be between 0.0 and 1.0")
        
    numerator = x ** (alpha - 1.0) * (1.0 - x) ** (beta - 1.0)
    denominator = (1.0 * factorial(alpha - 1) * factorial(beta - 1)) / (1.0 * factorial(alpha + beta - 1))
    
    return numerator / denominator

greater_than_90 = approximate_integral(a=0.9, b=1.0, n=1000, f=lambda x: beta_distribution(x, 8, 2))
less_than_90 = 1.0 - greater_than_90

print(f"Greater than 90%: {greater_than_90}, less than 90%: {less_than_90}")
```
