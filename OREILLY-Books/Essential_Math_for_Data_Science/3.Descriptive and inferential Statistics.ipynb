# Descriptive and inferential Statistics

### Mean

```python
sample = [1,3,2,5,7,0,2,3]
mean = sum(sample)/len(sample)

print(mean)
```

```python
#Wheigt mean

sample = [90,80,63,87,77]
weights = [0.20,0.10,0.20,0.40,0.10]

weight_mean = sum(s * w for s,w in zip(sample,weights)) / sum(weights)

print(weight_mean)
```

### Median

```python
sample = [0,1,14,5,7,9,10]

def median(values):
    ordered = sorted(values)
    print(ordered)
    
    n = len(ordered)
    
    if n%2 == 0:
        return (ordered[int(n/2)] + ordered[int(n/2 +1)]) / 2
    else:
        return (ordered[int(n/2)])
    
median(sample)
```
