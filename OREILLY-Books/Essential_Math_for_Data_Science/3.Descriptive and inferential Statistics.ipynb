# Descriptive and inferential Statistics

### Mean

```python
sample = [1,3,2,5,7,0,2,3]
mean = sum(sample)/len(sample)

print(mean)
```

```python
#Wheigt mean

sample = [90,80,63,87,77]
weights = [0.20,0.10,0.20,0.40,0.10]

weight_mean = sum(s * w for s,w in zip(sample,weights)) / sum(weights)

print(weight_mean)
```

### Median

```python
sample = [0,1,14,5,7,9,10]

def median(values):
    ordered = sorted(values)
    print(ordered)
    
    n = len(ordered)
    
    if n%2 == 0:
        return (ordered[int(n/2)] + ordered[int(n/2 +1)]) / 2
    else:
        return (ordered[int(n/2)])
    
median(sample)
```

### Mode

```python
#Number of pets each person owns]
from collections import defaultdict
```

```python
sample = [1,3,2,5,7,0,2,3]

def mode(values):
    counts = defaultdict(lambda:0)
    
    for s in values:
        counts[s] += 1
        
    max_count = max(counts.values())
    modes = [v for v in set(values) if counts[v] == max_count]
    return modes

print(mode(sample))
```

### Variance and Standard Deviation 

```python
#population variance = ((x1-mean)**2 +...+(xn-mean)**2) / N
```

```python
data = [0,1,5,7,9,10,14]

def variance(values):
    mean = sum(values) / len(values)
    _variance = sum((v - mean)**2 for v in values) / len (values)
    return _variance

print(variance(data))
```

```python
#square root of the variance give us standard deviation. This is the variance scaled into a number expressedin terms of"no. of pets"
```

```python
import math 
from math import sqrt
variance(data)
```

```python
math.sqrt(variance(data))
```

```python
#Sample variance = ((xi-mean)**2 +...+(xn-mean)**2) / (n -1)
```

```python
data = [0,1,5,7,9,10,14]

def variance(values, is_sample: bool=False):
    mean = sum(values) / len(values)
    _variance = sum((v - mean)**2 for v in values) / (len (values) - (1 if is_sample else 0))
    return _variance

def std_dev(values, is_sample: bool =False):
    return sqrt(variance(values, is_sample))

print("VARIANCE = {}".format(variance(data, is_sample=True)))
print("STD DEV = {}".format(std_dev(data, is_sample=True)))
```

### The Normal Distribution  

#### Probability Density Function (PDF) 

```python
# f(x) = (1 / (std_dev* sqrt(2*pi))) * e**(1/2((x-mu**2)std_dev))

#Defines gow spread out is the distribution
```

```python
def normal_pdf(x: float, mean:float, std_dev:float) -> float:
    return(1.0/(2.0 * math.pi * std_dev ** 2) ** 0.5) * math.exp(-1.0 * ((x-mean)**2 / (2.0 * std_dev **2)))
```

#### Cumulative Distribution Function (CDF)

```python
# S-shaped curve (sigmoid curve), projects the area up to that range in the PDF
```

```python
from scipy.stats import norm

mean = 64.43
std_dev = 2.99

x = norm.cdf(64.43, mean, std_dev)

print(x)

# 50% of the population
```

```python
mean = 64.43
std_dev = 2.99

x = norm.cdf(66, mean, std_dev) - norm.cdf(62, mean, std_dev)

print(x)

#Area between 66 and 62 --- 49 % of population 
```

#### Inverse CDF 

```python
# Look up a probability and then return the corresponding x-value
```

```python
from scipy.stats import norm

x = norm.ppf(0.95, loc=64.43, scale=2.99)

print(x)

# 95% of population in 69 
```
