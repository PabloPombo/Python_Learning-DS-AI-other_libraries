### Median

```python
sample = [0,1,14,5,7,9,10]

def median(values):
    ordered = sorted(values)
    print(ordered)
    
    n = len(ordered)
    
    if n%2 == 0:
        return (ordered[int(n/2)] + ordered[int(n/2 +1)]) / 2
    else:
        return (ordered[int(n/2)])
    
median(sample)
```

### Mode

```python
#Number of pets each person owns]
from collections import defaultdict
```

```python
sample = [1,3,2,5,7,0,2,3]

def mode(values):
    counts = defaultdict(lambda:0)
    
    for s in values:
        counts[s] += 1
        
    max_count = max(counts.values())
    modes = [v for v in set(values) if counts[v] == max_count]
    return modes

print(mode(sample))
```

### Variance and Standard Deviation 

```python
#population variance = ((x1-mean)**2 +...+(xn-mean)**2) / N
```

```python
data = [0,1,5,7,9,10,14]

def variance(values):
    mean = sum(values) / len(values)
    _variance = sum((v - mean)**2 for v in values) / len (values)
    return _variance

print(variance(data))
```

```python
#Sample variance = ((xi-mean)**2 +...+(xn-mean)**2) / (n -1)
```

```python
data = [0,1,5,7,9,10,14]

def variance(values, is_sample: bool=False):
    mean = sum(values) / len(values)
    _variance = sum((v - mean)**2 for v in values) / (len (values) - (1 if is_sample else 0))
    return _variance

def std_dev(values, is_sample: bool =False):
    return sqrt(variance(values, is_sample))

print("VARIANCE = {}".format(variance(data, is_sample=True)))
print("STD DEV = {}".format(std_dev(data, is_sample=True)))
```

### The Normal Distribution  

#### Probability Density Function (PDF) 

```python
# f(x) = (1 / (std_dev* sqrt(2*pi))) * e**(1/2((x-mu**2)std_dev))

#Defines gow spread out is the distribution
```

```python
def normal_pdf(x: float, mean:float, std_dev:float) -> float:
    return(1.0/(2.0 * math.pi * std_dev ** 2) ** 0.5) * math.exp(-1.0 * ((x-mean)**2 / (2.0 * std_dev **2)))
```

#### Cumulative Distribution Function (CDF)

```python
from scipy.stats import norm

mean = 64.43
std_dev = 2.99

x = norm.cdf(64.43, mean, std_dev)

print(x)

# 50% of the population
```

#### Z-Scores 

```python
# z = (x - mean) / std_dev
#convert an x-value coming from given distribution with mean and std dev into Z-score
```

```python
def z_score(x, mean, std):
    return (x - mean) / std

def z_to_x(z, mean, std):
    return (z * std) + mean

mean = 140000
std_dev = 3000
x = 150000

z = z_score(x, mean, std_dev)
back_to_x = z_to_x(z, mean, std_dev)

print(f"Z-Score: {z}")
print(f"Back to X: {back_to_x}")
```

### Confidence Intervals 

```python
#A range calculation showing how confidently we beleive a sample mean falls in a range for the population mean
```

```python
from scipy.stats import norm
```

```python
# critical z value capturing 95% of probability at the center of the standar normal distribution

def critical_z_value(p):
    norm_dist = norm(loc=00.0, scale=1.0)
    left_tail_area = (1.0 - p) / 2.0
    upper_area = 1.0 - ((1.0 -p) / 2.0)
    return norm_dist.ppf(left_tail_area), norm_dist.ppf(upper_area)

print(critical_z_value(p=0.95))
```

```python
#Confidence Interval in Python

#Formula to get this margen: 
# E = -+ zc (sample_std_dev)/sqrt(n)


from math import sqrt
from scipy.stats import norm
```

```python
def critical_z_value(p):
    norm_dist = norm(loc=00.0, scale=1.0)
    left_tail_area = (1.0 - p) / 2.0
    upper_area = 1.0 - ((1.0 -p) / 2.0)
    return norm_dist.ppf(left_tail_area), norm_dist.ppf(upper_area)

def confidence_interval(p, sample_mean, sample_std, n):
    #Sample size must be greater than 30
    
    lower, upper = critical_z_value(p)
    lower_ci = lower * (sample_std / sqrt(n))
    upper_ci = upper * (sample_std / sqrt(n))
    
    return sample_mean + lower_ci, sample_mean + upper_ci

print(confidence_interval(p=0.95, sample_mean=64.408, sample_std=2.05, n=31))
```

#### P-Value 

```python
#The probability of something occurring by chance rather than because of a hypothesized explanation
# H0: Null Hypothesis saying that the variable in question had no impact on the experiment and any 
#     positive results are just random luck
# H1: Alternative hypothesis poses that a variable in question (called the controlled variable) is 
#     causing a positive result
```

#### Hypothesis Testing 

```python
from scipy.stats import norm
```

```python
#Cold has 18 day mean recovery, 1.5 std dev
mean = 18
std_dev = 1.5

#95% probability recovery time takes between 15 and 21 days
x = norm.cdf(21, mean, std_dev) - norm.cdf(15, mean, std_dev)

print(x)
```

#### One-Tailed Test 

```python
# H0: population mean _> 18
# H1: population mean <  18
```

```python
#Cold has 18 day mean recovery, 1.5 std dev 
mean = 18
std_dev = 1.5

#What x-value has 5% of area behind it?
x = norm.ppf(0.05, mean, std_dev)

print(x)
```

```python
#Cold has 18 day mean recovery, 1.5 std dev 
mean = 18
std_dev = 1.5

#Probability of 16 or lessdays
x = norm.cdf(16, mean, std_dev)

print(x)
```

```python
#p-value of 0.0912 greater than our statistical significance threshold of 0.05 - fail to reject our null hypothesis
```

#### Two-Tailed Test 

```python
# H0: population mean = 18
# H1: population mean =! 18
```

```python
#Calculating Range of a statistical significance of 5%

#Cold has 18 day mean recovery, 1.5 std dev 
mean = 18
std_dev = 1.5

#What x -value has 2.5% of area behind it
x1 = norm.ppf(0.025, mean, std_dev)

#What x -value has 97.5% of area behind it
x2 = norm.ppf(0.975, mean, std_dev)

print(x1)
print(x2)
```

```python
#Still fail to reject the null hypothesis, the sample mean value for the drug test group is 16, and 16 is not less 
# than 15.06 nor greater than 20.9399
```

```python
#Calculating two-tailed p-value

#Cold has 18 day mean recovery, 1.5 std dev 
mean = 18
std_dev = 1.5

#Probability of 16 or less days
p1 = norm.cdf(16, mean, std_dev)

#Probability of 20 or more days
p2 = 1 - norm.cdf(20, mean, std_dev)

#P-value of both tails
p_value = p1+p2
print(p_value)
```

### CDF and inverse CDF from Scratch 

```python
import math

def normal_pdf(x: float, mean:float, std_dev:float) -> float:
    return(1.0/(2.0 * math.pi * std_dev ** 2) ** 0.5) * math.exp(-1.0 * ((x-mean)**2 / (2.0 * std_dev **2)))

def aproximate_integral(a, b, n, f):
    delta_x = (b - a) / n
    total_sum = 0
    
    for i in range (1, n + 1):
        midpoint = 0.5 * (2 * a + delta_x * (2 * i - 1))
        total_sum += f(midpoint)
        
    return total_sum *  delta_x

p_between_61_and_62 = aproximate_integral(a=61, b=62, n=7, f= lambda x: normal_pdf(x,64.43,2.99))

print(p_between_61_and_62)
```

```python
import math

def normal_cdf(x: float, mean:float, std_dev:float) -> float:
    return(1.0 + math.erf((x - mean) / math.sqrt(2) / std_dev)) / 2

mean = 64.43
std_dev = 2.99

x = normal_cdf(66,mean,std_dev) - normal_cdf(62,mean,std_dev)
print(x)
```

```python
import random 
from scipy.special import erfinv

def inv_normal_cdf(p: float, mean: float, std_dev:float):
    return mean + (std_dev * (2.0 ** 0.5) * erfinv((2.0 * p) - 1.0))

mean = 64.43
std_dev = 2.99

for i in range (0,1000):
    random_p = random.uniform(0.0, 1.0)
    print(inv_normal_cdf(random_p, mean, std_dev))
```
