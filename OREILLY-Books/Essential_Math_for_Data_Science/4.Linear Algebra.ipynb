```python
#APPLYING ROTATION AND THEN SHEAT TO ANY VECTOR

#[1 1] [0 -1] [x]
# 0 1   1  0   y
```

```python
#Transformation1
i_hat1 = array([0,1])
j_hat1 = array([-1,0])
transform1 = array([i_hat1, j_hat1]).transpose()

#Transformation2
i_hat2 = array([1,0])
j_hat2 = array([1,1])
transform2 = array([i_hat2, j_hat2]).transpose()

#Combined
combined = transform2 @ transform1

print(f"Combined Matrix: {combined}")

#declare vector v 
v = array([1,2])

new_v = combined.dot(v)


print(new_v)
```

```python
#Tranformation in reverse

#Transformation1
i_hat1 = array([0,1])
j_hat1 = array([-1,0])
transform1 = array([i_hat1, j_hat1]).transpose()

#Transformation2
i_hat2 = array([1,0])
j_hat2 = array([1,1])
transform2 = array([i_hat2, j_hat2]).transpose()

#Combined
combined = transform1 @ transform2

print(f"Combined Matrix: {combined}")

#declare vector v 
v = array([1,2])

new_v = combined.dot(v)


print(new_v)
```

### Determinants 

```python
#Determinant = describe how much a sampled area in a vector space changes in scale with linear tranformations
```

```python
from numpy.linalg import det
from numpy import array
```

```python
i_hat = array([3,0])
j_hat = array([0,2])

basis = array([i_hat, j_hat]).transpose()

determinant = det(basis)

print(determinant)
```

```python
np.linalg.det(basis)
```

```python
i_hat = array([1,0])
j_hat = array([1,1])

basis = array([i_hat, j_hat]).transpose()

determinant = det(basis)

print(determinant)
```

```python
#Negative determinant - the orientation has flipped

i_hat = array([-2,1])
j_hat = array([1,2])

basis = array([i_hat, j_hat]).transpose()

determinant = det(basis)

print(determinant)
```

```python
#Determinant = 0 -- means all of the space has been squished into a lesser dimension
```

```python
#Linear dependence

i_hat = array([3,-1.5])
j_hat = array([-2,1])

basis = array([i_hat, j_hat]).transpose()

determinant = det(basis)

print(determinant)
```

### Eigenvectors and Eigenvalues 

```python
#braking up a matrix into components that are easier to work with in different machine learning tasks

#Av = lambda*v
```

```python
from numpy import array, diag
from numpy.linalg import eig, inv
```

```python
A = np.array ([
    [1,2],
    [4,5],
])

eigenvals, eigenvecs = eig(A)

print(f"Eigenvalues: {eigenvals}")
print(f"Eigenvectors: {eigenvecs}")
```

```python
# A = Q * L * Q**-1

#Q = eigenvectors
#L = Diagonal of Egenvalues
```

```python
Q = eigenvecs
R = inv(Q)

L = diag(eigenvals)
B = Q @ L @ R

print(f"Matrix A = {B}")
```
