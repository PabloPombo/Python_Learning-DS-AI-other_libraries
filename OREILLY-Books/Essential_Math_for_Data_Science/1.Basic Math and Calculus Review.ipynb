### Summations

```python
i, n = symbols ('i,n')
#iterate each element i from 1 to n
#then multiply and sum
summation = Sum(2*i,(i,1,n))

#specify n as 5 
#iterating the numbers 1 through 5
up_to_5 = summation.subs(n, 5)
print(up_to_5.doit())
```

```python
x = symbols ('x')
expr = x**2 / x**5
print(expr)
```

### Euler's
